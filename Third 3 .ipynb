```python
def resolve(c1, c2):
    for lit1 in c1:
        for lit2 in c2:
            if lit1 == "~" + lit2 or "~" + lit1 == lit2:
                return list(set(c1 + c2) - {lit1, lit2})
    return None


def resolution(kb, query):
    clauses = kb + [["~" + q for q in query]]

    while True:
        new_clauses = []

        for i in range(len(clauses)):
            for j in range(i + 1, len(clauses)):
                resolvent = resolve(clauses[i], clauses[j])

                if resolvent is not None:
                    if not resolvent:
                        return True

                    if resolvent not in clauses and resolvent not in new_clauses:
                        new_clauses.append(resolvent)

        if not new_clauses:
            return False

        clauses.extend(new_clauses)


# Example
kb = [
    ["P(a)"],
    ["~P(a)", "Q(a)"]
]

query = ["Q(a)"]

if resolution(kb, query):
    print("Query is entailed by the knowledge base.")
else:
    print("Query is not entailed by the knowledge base.")
```
